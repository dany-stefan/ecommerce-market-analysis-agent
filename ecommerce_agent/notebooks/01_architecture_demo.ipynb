# Question 1: Base Architecture Demonstration

This notebook demonstrates the foundational architecture of the Market Analysis Agent.

## Architecture Overview

Our architecture consists of:
1. **BaseTool**: Abstract interface all tools implement
2. **Data Models**: Pydantic models for type safety
3. **MarketAnalysisAgent**: Orchestrator that coordinates tools

### Design Choice: Custom Implementation

We chose a **custom implementation** over frameworks like LangChain because:
- **Simplicity**: Easier to understand and explain
- **Transparency**: Clear control flow and debugging
- **Flexibility**: No framework constraints
- **Learning**: Better for presentations and demonstrations

```python
# Setup imports
import sys
sys.path.append('..')

from src.tools.base_tool import BaseTool, ToolInput, ToolOutput
from src.agent.orchestrator import MarketAnalysisAgent
from src.utils.models import AnalysisRequest, ProductData
from pydantic import BaseModel
```

## 1. Base Tool Architecture

Every tool inherits from `BaseTool` and implements two things:
- `description`: What the tool does
- `execute()`: The actual logic

```python
# Example: Create a simple mock tool
class MockToolInput(ToolInput):
    query: str

class SimpleMockTool(BaseTool):
    """A simple example tool for demonstration"""
    
    @property
    def description(self) -> str:
        return "A mock tool that returns test data"
    
    def execute(self, input_data: MockToolInput) -> ToolOutput:
        return ToolOutput(
            success=True,
            data={"message": f"Processed: {input_data.query}", "result": 42}
        )

# Test the tool
tool = SimpleMockTool()
print(f"Tool Name: {tool.name}")
print(f"Tool Description: {tool.description}")

# Run the tool
result = tool.run(MockToolInput(query="test query"))
print(f"\nTool Output:")
print(f"Success: {result.success}")
print(f"Data: {result.data}")
```

## 2. Data Models

Pydantic models ensure consistent data structures across tools.

```python
# Create a sample product data structure
product = ProductData(
    name="iPhone 15",
    price=999.00,
    currency="USD",
    description="Latest Apple smartphone",
    specifications={"storage": "128GB", "color": "Black"},
    source="mock_api"
)

print("Product Data Model:")
print(product.model_dump_json(indent=2))
```

## 3. Agent Orchestration

The agent coordinates tools in a simple, transparent way.

```python
# Initialize the agent
agent = MarketAnalysisAgent()

# Register our mock tool
agent.register_tool(SimpleMockTool())

# List available tools
print("Available tools:")
for tool_name in agent.list_tools():
    print(f"  - {tool_name}")
```

## 4. API Interface Design

Simple, intuitive API for running analyses.

```python
# Create an analysis request
request = AnalysisRequest(
    product_query="iPhone 15",
    analysis_depth="standard",
    include_competitors=True,
    include_sentiment=True
)

print("Analysis Request:")
print(request.model_dump_json(indent=2))
```
