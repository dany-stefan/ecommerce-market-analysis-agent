## Visualization Data Structures

The Report Generator creates **visualization-ready data** that can be easily consumed by any charting library (matplotlib, plotly, Chart.js, etc.).

Let's examine the visualization data structures:

```python
# Run complete analysis first
agent = MarketAnalysisAgent()
agent.register_tool(ProductCollectorTool(use_mock_data=True))
agent.register_tool(SentimentAnalyzerTool(use_llm=False))
agent.register_tool(ReportGeneratorTool(use_llm=False))

request = AnalysisRequest(
    product_query="iPhone 15 Pro",
    analysis_depth="comprehensive",
    include_competitors=True,
    include_sentiment=True
)

print("🔄 Running complete market analysis...\n")
analysis_result = agent.analyze(request)

# Generate report
report_tool = ReportGeneratorTool(use_llm=False)
report_result = report_tool.run(ReportGeneratorInput(
    analysis_result=analysis_result.model_dump()
))

print("="*70)
print("REPORT GENERATION RESULTS")
print("="*70)

if report_result.success:
    data = report_result.data
    
    print(f"\n📊 RECOMMENDATIONS ({len(data['recommendations'])}):")
    for i, rec in enumerate(data['recommendations'], 1):
        print(f"  {i}. {rec}")
    
    # Show visualization types
    print(f"\n📈 VISUALIZATIONS GENERATED ({len(data.get('visualizations', {}))}):") 
    for viz_name, viz_data in data.get('visualizations', {}).items():
        print(f"  • {viz_name}: {viz_data['chart_type']} - {viz_data['title']}")
    
    # Show markdown report preview
    print(f"\n📄 MARKDOWN REPORT (preview):")
    print("-"*70)
    print(data['markdown_report'][:500] + "...")
    print("-"*70)
    
    # Save full report
    os.makedirs("../reports", exist_ok=True)
    with open("../reports/question_2_demo_report.md", "w") as f:
        f.write(data['markdown_report'])
    print(f"\n✅ Full report saved to: reports/question_2_demo_report.md")
else:
    print(f"❌ Error: {report_result.error}")
```

## Tool 3: Report Generator with Visualizations

**Purpose:** Synthesizes analysis data into comprehensive reports with visualization structures

Let's generate a complete report:

```python
# Initialize Market Trend Analyzer
trend_tool = MarketTrendAnalyzerTool(use_mock_data=True)

# Run trend analysis for past 90 days
input_data = MarketTrendInput(
    product_name="iPhone 15 Pro",
    time_period_days=90,
    include_competitors=True
)

result = trend_tool.run(input_data)

# Display results
print("="*70)
print("MARKET TREND ANALYSIS RESULTS")
print("="*70)
print(f"\n✅ Analysis Status: {'SUCCESS' if result.success else 'FAILED'}\n")

if result.success:
    data = result.data
    print(f"📊 Product: {data['product_name']}")
    print(f"📅 Analysis Period: {data['analysis_period_days']} days")
    print(f"\n💰 PRICING TRENDS:")
    print(f"  • Trend Direction: {data['price_trend'].upper()}")
    print(f"  • Price Change: {data['price_change_percent']:+.2f}%")
    
    print(f"\n📈 POPULARITY TRENDS:")
    print(f"  • Trend Direction: {data['popularity_trend'].upper()}")
    print(f"  • Search Volume Change: {data['search_volume_change_percent']:+.2f}%")
    
    print(f"\n🎯 MARKET MOMENTUM: {data['current_momentum'].upper()}")
    
    print(f"\n🔮 FORECAST:")
    print(f"  {data['forecast_summary']}")
    
    # Show recent price points
    print(f"\n💵 Recent Price History (last 7 days):")
    for price_point in data['price_history'][-7:]:
        print(f"  • {price_point['date']}: ${price_point['price']:.2f}")
    
    # Show competitor comparison
    if data.get('competitor_comparison'):
        comp_data = data['competitor_comparison']
        print(f"\n🏆 COMPETITIVE LANDSCAPE:")
        print(f"  Target Product: {comp_data['target_product']['name']} - ${comp_data['target_product']['avg_price']:.2f}")
        print(f"  Market Share Estimate: {comp_data['market_share_estimate']}")
        print(f"\n  Top Competitors:")
        for i, comp in enumerate(comp_data['competitors'][:3], 1):
            print(f"    {i}. {comp['name']}: ${comp['avg_price']:.2f} ({comp['price_vs_target']}) - {comp['market_position']}")
else:
    print(f"❌ Error: {result.error}")
```

## Tool 2: Market Trend Analyzer ⭐ NEW

**Purpose:** Analyzes price and popularity trends over time for market intelligence

**Key Features:**
- Historical price tracking with trend detection
- Popularity metrics (search volume, review counts, ratings)
- Market momentum indicators (bullish, bearish, opportunity, warning)
- Competitive trend comparison  
- Actionable forecast insights

Let's analyze market trends:

```python
# Initialize Sentiment Analyzer
sentiment_tool = SentimentAnalyzerTool(use_llm=False)  # Change to True if you have OpenAI API key

# Sample reviews
reviews = [
    "The iPhone 15 Pro has an amazing camera system!",
    "Battery life could be better, but overall great phone",
    "Love the titanium build quality",
    "Expensive but worth every penny",
    "Best smartphone I've ever owned",
    "Camera quality is absolutely stunning",
    "The new Action button is very convenient",
    "Charging is slower than competitors"
]

# Run sentiment analysis
input_data = SentimentAnalyzerInput(
    product_name="iPhone 15 Pro",
    reviews=reviews
)

result = sentiment_tool.run(input_data)

# Display results
print("="*70)
print("SENTIMENT ANALYSIS RESULTS")
print("="*70)
print(f"\n✅ Analysis Status: {'SUCCESS' if result.success else 'FAILED'}\n")

if result.success:
    data = result.data
    print(f"📊 Overall Sentiment: {data['overall_sentiment'].upper()}")
    print(f"📈 Sentiment Score: {data['sentiment_score']:.2f} / 1.0")
    print(f"📝 Total Reviews Analyzed: {data['total_reviews']}")
    print(f"\n🔑 Key Themes:")
    for i, theme in enumerate(data['key_themes'], 1):
        print(f"  {i}. {theme}")
    print(f"\n💬 Sample Representative Reviews:")
    for i, review in enumerate(data['sample_reviews'][:3], 1):
        print(f"  {i}. \"{review}\"")
else:
    print(f"❌ Error: {result.error}")
```
