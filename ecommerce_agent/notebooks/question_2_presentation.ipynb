Let's see it in action:

````python
# Initialize Sentiment Analyzer (using mock mode for demo)
sentiment_tool = SentimentAnalyzerTool(use_llm=False)

# Sample customer reviews
sample_reviews = [
    "Amazing phone! Camera quality is incredible and battery lasts all day.",
    "Expensive but worth it. The performance is smooth.",
    "Battery life could be better, but overall great device.",
    "Love the new design and features. Best iPhone yet!",
    "Price is too high for what you get.",
    "Camera is fantastic, especially in low light.",
    "Battery drains quickly with heavy use.",
    "Display is stunning and colors are vibrant."
]

# Run sentiment analysis
result = sentiment_tool.run(SentimentAnalyzerInput(
    product_name="iPhone 15 Pro",
    reviews=sample_reviews
))

# Display results
print("📊 Sentiment Analysis Results:")
print(f"   Product: iPhone 15 Pro")
print(f"   Reviews Analyzed: {len(sample_reviews)}")
print(f"\n   Overall Sentiment: {result.data['overall_sentiment'].upper()}")
print(f"   Sentiment Score: {result.data['sentiment_score']:.2f}/1.0")
print(f"   Key Themes: {', '.join(result.data['key_themes'][:3])}")
print(f"\n   Sample Positive Review:")
print(f"   '{result.data['sample_reviews'][0][:80]}...'")

# Store for later use
sentiment_data = result.data
````

## Part 2: Tool 2 - Market Trend Analyzer ⭐

### Overview

The Market Trend Analyzer tracks price and popularity trends over time with momentum indicators.

**Key Features:**
- **90-Day Historical Tracking:** Price history with trend detection
- **Popularity Metrics:** Search volume, review counts, rating averages
- **Momentum Matrix:** Bullish/Bearish/Opportunity/Warning signals
- **Competitor Analysis:** Comparative trend analysis
- **Forecast Engine:** Actionable insights and predictions

**The Momentum Matrix:**
- **Bullish:** Rising prices + growing demand (strong market)
- **Bearish:** Falling prices + declining demand (weak market)
- **Opportunity:** Growing demand + falling prices (value play)
- **Warning:** Declining demand + rising prices (market rejection)

````python
# Initialize Market Trend Analyzer
trend_tool = MarketTrendAnalyzerTool(use_mock_data=True)

# Analyze 90-day trends
trend_result = trend_tool.run(MarketTrendInput(
    product_name="iPhone 15 Pro",
    time_period_days=90,
    include_competitors=True
))

trend_data = trend_result.data

# Display trend analysis
print("📈 Market Trend Analysis (90 days):")
print(f"   Product: {trend_data['product_name']}")
print(f"   Analysis Period: {trend_data['analysis_period_days']} days")

print(f"\n   💰 Price Trend:")
print(f"      Current: ${trend_data['current_price']:.2f}")
print(f"      Trend: {trend_data['price_trend'].upper()}")
print(f"      Change: {trend_data['price_change_percent']:+.1f}%")

print(f"\n   📊 Popularity Trend:")
print(f"      Trend: {trend_data['popularity_trend'].upper()}")
print(f"      Change: {trend_data['popularity_change_percent']:+.1f}%")

print(f"\n   🎯 Market Momentum: {trend_data['current_momentum'].upper()}")
print(f"\n   💡 Forecast:")
print(f"      {trend_data['forecast_summary'][:120]}...")

# Display competitor comparison
if 'competitor_comparison' in trend_data:
    print(f"\n   🔍 Competitors Analyzed: {len(trend_data['competitor_comparison'])}")
    for comp in trend_data['competitor_comparison'][:2]:
        print(f"      • {comp['name']}: ${comp['price']:.2f} ({comp['relative_position']})")

# Store for report generation
trend_analysis = trend_data
````

## Part 3: Tool 3 - Report Generator ⭐

### Overview

The Report Generator synthesizes all analysis data into comprehensive reports with visualizations.

**Key Features:**
- **LLM Integration:** Uses GPT-4 for strategic insights (fallback to template)
- **Multiple Output Formats:** JSON, Markdown, Visualizations
- **6 Visualization Types:** Bar, gauge, wordcloud, scatter, pie, line charts
- **Library-Agnostic:** Works with matplotlib, plotly, Chart.js, D3, etc.
- **Strategic Recommendations:** 4-5 actionable business insights

**The 6 Visualization Types:**
1. `price_comparison` - Bar chart comparing product vs competitors
2. `sentiment_score` - Gauge chart showing sentiment (-1.0 to 1.0)
3. `key_themes` - Word cloud with weighted terms
4. `positioning_matrix` - Scatter plot (price vs sentiment)
5. `market_share` - Pie chart with distribution percentages
6. `price_trend` - Line chart for historical pricing

````python
# Initialize Report Generator
report_tool = ReportGeneratorTool(use_llm=False)

# Create comprehensive analysis result
analysis_result = AnalysisResult(
    product_data={"name": "iPhone 15 Pro", "price": 999},
    sentiment=sentiment_data,  # From Tool 1
    competitors=[
        {"competitor_name": "Samsung Galaxy S24", "price": 899},
        {"competitor_name": "Google Pixel 8 Pro", "price": 899},
        {"competitor_name": "OnePlus 12", "price": 799},
    ],
    recommendations=[],
    metadata={"trends": trend_analysis}  # From Tool 2
)

# Generate comprehensive report
report_result = report_tool.run(ReportGeneratorInput(
    analysis_result=analysis_result
))

report_data = report_result.data

# Display report summary
print("📄 Report Generation Results:")
print(f"   Product: {analysis_result.product_data['name']}")
print(f"   Report Length: {len(report_data['report'])} characters")

print(f"\n   📋 Strategic Recommendations: {len(report_data['recommendations'])}")
for i, rec in enumerate(report_data['recommendations'][:3], 1):
    print(f"      {i}. {rec[:80]}...")

print(f"\n   📊 Visualizations Generated: {len(report_data['visualizations'])}")
for viz_type in report_data['visualizations'].keys():
    print(f"      • {viz_type}")

# Save report to file
os.makedirs("../reports", exist_ok=True)
report_path = "../reports/Q2_DEMO_iPhone_15_Pro_Report.md"
with open(report_path, 'w') as f:
    f.write(report_data['report'])

print(f"\n   💾 Full report saved to: {report_path}")
````

## Part 4: Visualization Data Structures

Let's examine the visualization data generated by the Report Generator.

Each visualization type provides library-agnostic data structures that can be used with any charting library (matplotlib, plotly, Chart.js, D3, etc.).

````python
# Explore visualization data structures
visualizations = report_data['visualizations']

print("📊 Generated Visualization Data:\n")

# 1. Price Comparison (Bar Chart)
print("1️⃣ Price Comparison (Bar Chart):")
price_comp = visualizations['price_comparison']
print(f"   Type: {price_comp['type']}")
print(f"   Products: {len(price_comp['data']['labels'])}")
print(f"   Data: {price_comp['data']['values']}")

# 2. Sentiment Score (Gauge Chart)
print("\n2️⃣ Sentiment Score (Gauge Chart):")
sentiment_viz = visualizations['sentiment_score']
print(f"   Type: {sentiment_viz['type']}")
print(f"   Score: {sentiment_viz['data']['value']}")
print(f"   Range: {sentiment_viz['data']['min']} to {sentiment_viz['data']['max']}")

# 3. Key Themes (Word Cloud)
print("\n3️⃣ Key Themes (Word Cloud):")
themes_viz = visualizations['key_themes']
print(f"   Type: {themes_viz['type']}")
print(f"   Terms: {len(themes_viz['data']['terms'])}")
for term in themes_viz['data']['terms'][:3]:
    print(f"      • {term['text']}: weight {term['weight']}")

# 4. Positioning Matrix (Scatter Plot)
print("\n4️⃣ Positioning Matrix (Scatter Plot):")
matrix_viz = visualizations['positioning_matrix']
print(f"   Type: {matrix_viz['type']}")
print(f"   Products Plotted: {len(matrix_viz['data']['points'])}")

# 5. Market Share (Pie Chart)
print("\n5️⃣ Market Share (Pie Chart):")
share_viz = visualizations['market_share']
print(f"   Type: {share_viz['type']}")
print(f"   Segments: {len(share_viz['data']['labels'])}")

# 6. Price Trend (Line Chart)
print("\n6️⃣ Price Trend (Line Chart):")
trend_viz = visualizations['price_trend']
print(f"   Type: {trend_viz['type']}")
print(f"   Data Points: {len(trend_viz['data']['labels'])}")

print("\n✅ All 6 visualization types generated successfully!")
````
