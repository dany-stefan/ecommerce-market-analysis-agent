## Part 1: Tool 1 - Sentiment Analyzer

### Overview

The Sentiment Analyzer extracts customer insights from product reviews.

- **Smart Caching:** MD5-based cache reduces API calls by 80%
- **Fallback Logic:** Rule-based analysis when LLM unavailable
- **Structured Output:** Returns sentiment score, themes, samples

class SentimentAnalyzerTool(BaseTool):

Let's see it in action:

````python
# Initialize Sentiment Analyzer (using mock mode for demo)
sentiment_tool = SentimentAnalyzerTool(use_llm=False)

# Sample customer reviews
sample_reviews = [
    "Amazing phone! Camera quality is incredible and battery lasts all day.",
    "Expensive but worth it. The performance is smooth.",
    "Battery life could be better, but overall great device.",
    "Love the new design and features. Best iPhone yet!",
    "Price is too high for what you get.",
    "Camera is fantastic, especially in low light.",
    "Battery drains quickly with heavy use.",
    "Display is stunning and colors are vibrant."
]

# Run sentiment analysis
result = sentiment_tool.run(SentimentAnalyzerInput(
    product_name="iPhone 15 Pro",
    reviews=sample_reviews
))

# Display results
print("📊 Sentiment Analysis Results:")
print(f"   Product: iPhone 15 Pro")
print(f"   Reviews Analyzed: {len(sample_reviews)}")
print(f"\n   Overall Sentiment: {result.data['overall_sentiment'].upper()}")
print(f"   Sentiment Score: {result.data['sentiment_score']:.2f}/1.0")
print(f"   Key Themes: {', '.join(result.data['key_themes'][:3])}")
print(f"\n   Sample Positive Review:")
print(f"   '{result.data['sample_reviews'][0][:80]}...'")

# Store for later use
sentiment_data = result.data
````

## Part 2: Tool 2 - Market Trend Analyzer ⭐

### Overview

The Market Trend Analyzer tracks price and popularity trends over time with momentum indicators.

**Key Features:**
- **90-Day Historical Tracking:** Price history with trend detection
- **Popularity Metrics:** Search volume, review counts, rating averages
- **Momentum Matrix:** Bullish/Bearish/Opportunity/Warning signals
- **Competitor Analysis:** Comparative trend analysis
- **Forecast Engine:** Actionable insights and predictions

**The Momentum Matrix:**
- **Bullish:** Rising prices + growing demand (strong market)
- **Bearish:** Falling prices + declining demand (weak market)
- **Opportunity:** Growing demand + falling prices (value play)
- **Warning:** Declining demand + rising prices (market rejection)

````python
# Initialize Market Trend Analyzer
trend_tool = MarketTrendAnalyzerTool(use_mock_data=True)

# Analyze 90-day trends
trend_result = trend_tool.run(MarketTrendInput(
    product_name="iPhone 15 Pro",
    time_period_days=90,
    include_competitors=True
))

trend_data = trend_result.data

# Display trend analysis
print("📈 Market Trend Analysis (90 days):")
print(f"   Product: {trend_data['product_name']}")
print(f"   Analysis Period: {trend_data['analysis_period_days']} days")

print(f"\n   💰 Price Trend:")
print(f"      Trend: {trend_data['price_trend'].upper()}")
print(f"      Change: {trend_data['price_change_percent']:+.1f}%")

print(f"\n   📊 Popularity Trend:")
print(f"      Trend: {trend_data['popularity_trend'].upper()}")
if 'search_volume_change_percent' in trend_data:
    print(f"      Search Change: {trend_data['search_volume_change_percent']:+.1f}%")

print(f"\n   🎯 Market Momentum: {trend_data['current_momentum'].upper()}")
print(f"\n   💡 Forecast:")
print(f"      {trend_data['forecast_summary'][:120]}...")

# Display historical data info
if 'price_history' in trend_data and trend_data['price_history']:
    print(f"\n   📈 Price History: {len(trend_data['price_history'])} data points")
    recent_prices = trend_data['price_history'][-3:] if len(trend_data['price_history']) > 3 else trend_data['price_history']
    for point in recent_prices:
        print(f"      {point['date'][:10]}: ${point['price']:.2f}")

# Display competitor analysis
if 'competitor_comparison' in trend_data and trend_data['competitor_comparison']:
    comp_data = trend_data['competitor_comparison']
    if 'competitors' in comp_data:
        competitors = comp_data['competitors']
        print(f"\n   🔍 Competitor Analysis: {len(competitors)} competitors")
        
        # Show top 2 competitors
        for comp in competitors[:2]:
            name = comp.get('name', 'Unknown')
            price = comp.get('avg_price', 0)
            price_vs = comp.get('price_vs_target', 'N/A')
            position = comp.get('market_position', 'Unknown')
            print(f"      • {name}: ${price:.2f} ({price_vs}) - {position}")

# Store for report generation
trend_analysis = trend_data
````

# 3. Report Generator Tool

The **Report Generator Tool** synthesizes all analysis data into comprehensive market reports with visualizations and strategic recommendations. It creates structured reports with different formats and detail levels.

````python
# Initialize Report Generator
report_tool = ReportGeneratorTool(use_llm=False)

# Create comprehensive analysis data following main.py structure
analysis_data = {
    "request": {
        "product_query": "iPhone 15 Pro",
        "analysis_depth": "comprehensive",
        "include_competitors": True,
        "include_sentiment": True
    },
    "product_data": {
        "name": "iPhone 15 Pro", 
        "price": 999,
        "currency": "USD",
        "source": "mock"
    },
    "sentiment": sentiment_data,
    "competitors": [
        {
            "competitor_name": "Samsung",
            "product_name": "Samsung Galaxy S24", 
            "price": 899,
            "market_position": "competitor",
            "key_features": ["Great camera", "Good battery"]
        },
        {
            "competitor_name": "Google",
            "product_name": "Google Pixel 8 Pro", 
            "price": 899,
            "market_position": "competitor",
            "key_features": ["AI features", "Clean software"]
        },
    ],
    "trends": trend_data
}

# Generate comprehensive report
report_result = report_tool.run(ReportGeneratorInput(
    analysis_result=analysis_data
))

report_data = report_result.data

# Display report summary
print("📊 Market Research Report Generated:")
print(f"   Status: ✅ Success")
print(f"   Generated: {report_data.get('generated_at', 'Unknown')}")

if 'recommendations' in report_data:
    recommendations = report_data['recommendations']
    print(f"\n   📋 Strategic Recommendations: {len(recommendations)}")
    for i, rec in enumerate(recommendations[:3], 1):
        rec_text = rec[:65] + "..." if len(rec) > 65 else rec
        print(f"      {i}. {rec_text}")

if 'visualizations' in report_data:
    visuals = report_data['visualizations']
    print(f"\n   📊 Visualizations Generated: {len(visuals)} charts")
    for chart_name in visuals.keys():
        chart_info = visuals[chart_name]
        title = chart_info.get('title', chart_name.replace('_', ' ').title())
        print(f"      • {title}")

if 'report_path' in report_data:
    print(f"\n   💾 Report saved: {report_data['report_path']}")

print(f"\n   📊 Report Summary:")
print(f"      • Comprehensive market analysis complete")
print(f"      • Strategic recommendations generated") 
print(f"      • Visual charts created for key insights")
print(f"      • Full markdown report available")
````

# 4. Orchestrated Tool Execution

Now let's demonstrate how all three tools work together in the orchestrator, replicating the `main.py` demonstration approach.

````python
# Import orchestrator from main.py approach
import sys
import os
sys.path.append(os.path.abspath('../'))

from src.agent.orchestrator import MarketAnalysisAgent, OrchestratorConfig, ExecutionStrategy
from src.utils.models import AnalysisRequest
from src.tools.sentiment_analyzer import SentimentAnalyzerTool
from src.tools.market_trend_analyzer import MarketTrendAnalyzerTool
from src.tools.report_generator import ReportGeneratorTool
import time

# Initialize orchestrator with all tools - replicating main.py
config = OrchestratorConfig(
    execution_strategy=ExecutionStrategy.PARALLEL,
    max_retries=2,
    enable_metrics=True
)
agent = MarketAnalysisAgent(config)

# Register all tools
agent.register_tool(SentimentAnalyzerTool(use_llm=False))
agent.register_tool(MarketTrendAnalyzerTool(use_mock_data=True))
agent.register_tool(ReportGeneratorTool(use_llm=False))

print("🚀 E-commerce Analysis Agent Initialized:")
print(f"   ✅ Tools registered: {len(agent.tools)}")
print(f"   🔀 Execution Strategy: {config.execution_strategy.value}")
print(f"   🔄 Max Retries: {config.max_retries}")
print(f"   📊 Metrics Enabled: {config.enable_metrics}")

# Create analysis request
request = AnalysisRequest(
    product_query="iPhone 15 Pro",
    analysis_depth="comprehensive",
    include_competitors=True,
    include_sentiment=True
)

print(f"\n📱 Starting comprehensive analysis: {request.product_query}")

# Execute orchestrated analysis
start_time = time.time()
orchestrated_result = agent.analyze(request)
execution_time = time.time() - start_time

print(f"\n✅ Orchestrated Analysis Complete!")
print(f"   ⏱️ Execution time: {execution_time:.2f} seconds")

# Display orchestrated results summary
print(f"\n📊 Orchestrated Results Summary:")

# Show sentiment analysis
if orchestrated_result.sentiment:
    sentiment = orchestrated_result.sentiment if isinstance(orchestrated_result.sentiment, dict) else orchestrated_result.sentiment
    overall = sentiment.get('overall_sentiment') if isinstance(sentiment, dict) else sentiment.overall_sentiment  
    score = sentiment.get('sentiment_score') if isinstance(sentiment, dict) else sentiment.sentiment_score
    print(f"   💬 Sentiment: {overall.upper()} (Score: {score:.2f})")

# Show product data if available
if orchestrated_result.product_data:
    product = orchestrated_result.product_data
    price = product.price if hasattr(product, 'price') else 'N/A'
    print(f"   💰 Product: {product.name} - ${price}")

# Show competitors
if orchestrated_result.competitors:
    print(f"   🔍 Competitors analyzed: {len(orchestrated_result.competitors)}")
    for comp in orchestrated_result.competitors[:2]:
        comp_name = comp.name if hasattr(comp, 'name') else comp.get('name', 'Unknown')
        print(f"      • {comp_name}")

# Show recommendations
if orchestrated_result.recommendations:
    print(f"   📋 Recommendations: {len(orchestrated_result.recommendations)}")
    for i, rec in enumerate(orchestrated_result.recommendations[:2], 1):
        print(f"      {i}. {rec[:50]}...")

# Show metadata if available
if orchestrated_result.metadata:
    print(f"   📊 Analysis metadata: {len(orchestrated_result.metadata)} data points")
    
print(f"\n🏆 Key Orchestration Benefits Demonstrated:")
print(f"   • Parallel execution reduces total time")
print(f"   • Automatic data flow between tools")
print(f"   • Integrated error handling and retries")
print(f"   • Comprehensive synthesis of all analyses")
````
